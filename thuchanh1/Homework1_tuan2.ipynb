```python
ten = input("Ten chu ho:")
so_dien_dau_thang = int(input("Chi so thang truoc:"))
so_dien_cuoi_thang = int(input("Chi so thang nay:"))

so_dien = so_dien_cuoi_thang - so_dien_dau_thang
tien_dien = 0

if so_dien > 0:
    tien_dien += min(so_dien, 50) * 1984
    so_dien -= min(so_dien, 50)
if so_dien > 0:
    tien_dien += min(so_dien, 50) * 2050
    so_dien -= min(so_dien, 50)
if so_dien > 0:
    tien_dien += min(so_dien, 100) * 2380
    so_dien -= min(so_dien, 100)
if so_dien > 0:
    tien_dien += min(so_dien, 100) * 2998
    so_dien -= min(so_dien, 100)
if so_dien > 0:
    tien_dien += min(so_dien, 100) * 3350
    so_dien -= min(so_dien, 100)
if so_dien > 0:
    tien_dien += so_dien * 3460

print(f"Ho va ten: {ten} ")
print(f"Tien phai tra la: {tien_dien*1.08:.0f}")
```
